```python
order_amounts = [120, 250, 90, 310, 150]

total = 0
for amount in order_amounts:
    total += amount

print("Total order value:", total)
```

```python
scores = [45, 78, 102, 34, 67, 89]

i = 0
while i < len(scores) and scores[i] <= 100:
    print(scores[i])
    i += 1
```

```python
prices = [299, 499, 199, 999, 149]

total = 0
for price in prices:
    if price < 200:
        continue
    total += price

print("Total of remaining items:", total)
```

```python
songs = ['Kesariya', 'Believer', 'Shape of You', 'Blinding Lights', 'Excuses']

for position, song in enumerate(songs, start=1):
    print(position, song)
```

```python
followers_list = [120, 1500, 23000, 800, 45000]

for followers in followers_list:
    if followers < 1000:
        print(followers, "- Micro")
    elif followers <= 10000:
        print(followers, "- Influencer")
    else:
        print(followers, "- Celebrity")
```
