```python
playlist_prices = {
    "Chill Vibes": 99,
    "Workout Beats": 149,
    "Bollywood Hits": 129,
    "Study Focus": 79,
    "Road Trip": 119
}

print(playlist_prices)
```

```python
def update_playlist_price(playlist, new_price):
    playlist_prices[playlist] = new_price


update_playlist_price("Chill Vibes", 109)

print(playlist_prices)
```

```python
del playlist_prices["Study Focus"]

print(playlist_prices)
```
