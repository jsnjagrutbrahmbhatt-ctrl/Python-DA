```python
import re

text = "Call me on 9876543210 or 8123456789. Price is 15000, pin 380001. Old number 6123456789 and 12345678901 are not valid. Also try 7000012345."

pattern = r'\b[789]\d{9}\b'

phones = re.findall(pattern, text)
print(phones)
```

```python
import re

def has_valid_date(text):
    pattern = r'\b\d{2}/\d{2}/\d{4}\b'
    if re.search(pattern, text):
        return True
    return False


print(has_valid_date("Order placed on 25/06/2024 by Rahul"))
print(has_valid_date("Order placed on 2024-06-25"))
print(has_valid_date("Date: 5/6/2024"))
```

```python
import re

review = """
Loved the food! Contact me at rahul.sharma@gmail.com or priya_99@yahoo.co.in.
For complaints write to support@zomato.com, not to fake@email or hello@.com.
Also try amit-k@outlook.com for catering.
"""

pattern = r'[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}'

emails = re.findall(pattern, review)
print(emails)
```

```python
import re

text = "Send money to 9876543210 or 8123456789 now."

masked = re.sub(r'\d(?=\d{4})', '*', text)
print(masked)
```

```python
import re

pattern = r'\bOD\d{18}\b'

orders = [
    "Your order OD123456789012345000 has been shipped",
    "Order ID: OD987654321098765432 delivered",
    "Invalid ID: OD12345 is too short",
    "Wrong prefix: XY123456789012345000",
    "No order here"
]

for order in orders:
    match = re.search(pattern, order)
    if match:
        print("Found:", match.group())
    else:
        print("No valid order ID in:", order)
```
