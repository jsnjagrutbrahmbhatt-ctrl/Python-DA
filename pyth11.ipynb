```python
def reverse_string(s):
    if len(s) <= 1:
        return s
    return reverse_string(s[1:]) + s[0]


print(reverse_string('hello'))
```

```python
def sum_playlist_durations(durations):
    if len(durations) == 0:
        return 0
    return durations[0] + sum_playlist_durations(durations[1:])


playlist = [210, 185, 240, 195]

total_seconds = sum_playlist_durations(playlist)
print("Total duration:", total_seconds, "seconds")
print("That is", total_seconds // 60, "minutes and", total_seconds % 60, "seconds")
```

```python
def count_likes(posts):
    total = 0
    for name, post in posts.items():
        total += post['likes']
        total += count_likes(post.get('replies', {}))
    return total


posts = {
    'post1': {
        'likes': 100,
        'replies': {
            'reply1': {
                'likes': 20,
                'replies': {
                    'reply1a': {'likes': 5}
                }
            },
            'reply2': {'likes': 10}
        }
    },
    'post2': {
        'likes': 50,
        'replies': {
            'reply3': {'likes': 8}
        }
    }
}

print("Total likes:", count_likes(posts))
```

```python
app_status = "Online"          # global variable

def change_status():
    user_status = "Typing..."  # local variable
    print("During call - user_status (local):", user_status)
    print("During call - app_status (global):", app_status)


print("Before call - app_status:", app_status)
try:
    print(user_status)
except NameError as e:
    print("Before call - user_status:", e)

change_status()

print("After call - app_status:", app_status)
try:
    print(user_status)
except NameError as e:
    print("After call - user_status:", e)
```
