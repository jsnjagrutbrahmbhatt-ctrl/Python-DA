```python
add_gst = lambda price: price * 1.18

print(add_gst(100))
print(add_gst(250))
print(add_gst(500))
```

```python
songs = [' shape OF you ', '  BLINDING lights', 'kesariya  ']

cleaned_songs = list(map(lambda s: s.strip().title(), songs))

print(cleaned_songs)
```

```python
products = ['Samsung Phone', 'laptop bag', 'Sony Headphones', 'shoes', 'Table Lamp', 'smart watch']

s_products = list(filter(lambda p: p.lower().startswith('s'), products))

print(s_products)
```

```python
from functools import reduce

order_amounts = [120, 340, 560, 80]

total_bill = reduce(lambda a, b: a + b, order_amounts)

print("Total bill amount:", total_bill)
```

```python
from functools import reduce

numbers = [40, 60, 80, 120]

doubled = list(map(lambda x: x * 2, numbers))
print("Doubled:", doubled)

filtered = list(filter(lambda x: x > 100, doubled))
print("Filtered:", filtered)

total = reduce(lambda a, b: a + b, filtered)
print("Sum:", total)
```
