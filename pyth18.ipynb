```python
class Product:
    def __init__(self, price):
        self._price = price

    def get_price(self):
        return self._price

    def set_price(self, new_price):
        if new_price < 0:
            raise ValueError("Price cannot be negative")
        self._price = new_price


item = Product(999)

print("Current price:", item.get_price())

item.set_price(799)
print("Updated price:", item.get_price())

try:
    item.set_price(-50)
except ValueError as e:
    print("Error:", e)
```

```python
class Playlist:
    def __init__(self):
        self._songs = []

    def add_song(self, song):
        self._songs.append(song)

    def remove_song(self, song):
        if song in self._songs:
            self._songs.remove(song)
        else:
            print(song, "is not in the playlist")

    def get_songs(self):
        return list(self._songs)


my_playlist = Playlist()

my_playlist.add_song("Kesariya")
my_playlist.add_song("Believer")
my_playlist.add_song("Excuses")
print("After adding:", my_playlist.get_songs())

my_playlist.remove_song("Believer")
print("After removing:", my_playlist.get_songs())

my_playlist.remove_song("Shape of You")
```

```python
from abc import ABC, abstractmethod


class PaymentMethod(ABC):
    @abstractmethod
    def pay(self, amount):
        pass


class UPI(PaymentMethod):
    def pay(self, amount):
        print(f"Processing ₹{amount} via UPI: verifying UPI ID and sending a collect request.")


class CreditCard(PaymentMethod):
    def pay(self, amount):
        print(f"Processing ₹{amount} via Credit Card: validating card number and OTP.")


upi_payment = UPI()
card_payment = CreditCard()

upi_payment.pay(500)
card_payment.pay(1200)

try:
    method = PaymentMethod()
except TypeError as e:
    print("Error:", e)
```
