```python
def remove_last_item(order_list):
    removed_item = order_list.pop()
    return removed_item


order_list = ['Biryani', 'Paneer Tikka', 'Naan', 'Lassi']

removed = remove_last_item(order_list)

print("Removed item:", removed)
print("Remaining order:", order_list)
```

```python
insta_filters = ('Clarendon', 'Gingham', 'Juno', 'Lark')

print(insta_filters)

try:
    insta_filters[1] = 'Moon'
except TypeError as e:
    print("Error:", e)

# This error happens because tuples are immutable, which means their items
# cannot be changed, added or removed after the tuple is created.
```

```python
# Scenario 1: user's favorite genres -> LIST
favorite_genres = ['Pop', 'Rock', 'Hip-Hop']

# Scenario 2: IRCTC train classes -> TUPLE
train_classes = ('Sleeper', 'AC 3 Tier', 'AC 2 Tier')

# Favorite genres should be a list because the user can add, remove or change them
# over time, while the IRCTC train classes should be a tuple because they are a fixed
# set that should never change.

print(favorite_genres)
print(train_classes)
```
