```python
import math

num = 16
n = 5

print("Square root of", num, "=", math.sqrt(num))
print("Factorial of", n, "=", math.factorial(n))
print("Value of pi =", math.pi)
```

```python
import os

files = os.listdir()

print("All files:", files)

for file in files:
    if file.lower().endswith((".jpg", ".png")):
        print(file)
```

```python
import datetime

date_text = input("Enter a date (YYYY-MM-DD): ")

date_obj = datetime.datetime.strptime(date_text, "%Y-%m-%d")

print("Day of the week:", date_obj.strftime("%A"))
```

```python
def format_follower_count(n):
    if n >= 1_000_000:
        return f"{n / 1_000_000:.1f}M"
    elif n >= 1_000:
        return f"{n / 1_000:.1f}K"
    else:
        return str(n)
```
