```python
price = input("Enter the price of your Zomato order: ")

price = float(price)             # convert string to float
final_bill = price + (price * 0.18)   # add 18% GST

print("Final bill amount:", final_bill)
```

```python
prices = ['199.99', '299.50', '150']

total = 0
for p in prices:
    total += float(p)

print("Total cart value:", round(total, 2))
```

```python
def is_discount_applicable(order_amount):
    return order_amount > 500

print(is_discount_applicable(450))
print(is_discount_applicable(750))
```

```python
ratings = ['4.5', '3.0', '5', '4.2']

ratings = [float(r) for r in ratings]

print("Highest rating:", max(ratings))
```
