```python
songs = ["Kesariya", "Believer", "Shape of You", "Blinding Lights", "Excuses"]

with open("my_playlist.txt", "w") as file:
    for song in songs:
        file.write(song + "\n")

print("File created: my_playlist.txt")
```

```python
with open("my_playlist.txt", "r") as file:
    for line in file:
        print(line.strip().upper())
```

```python
import csv

matches = [
    ["Match", "Team1", "Team2", "Winner"],
    [1, "Mumbai Indians", "Chennai Super Kings", "Chennai Super Kings"],
    [2, "Gujarat Titans", "Rajasthan Royals", "Gujarat Titans"],
    [3, "Royal Challengers Bengaluru", "Kolkata Knight Riders", "Kolkata Knight Riders"],
    [4, "Delhi Capitals", "Sunrisers Hyderabad", "Delhi Capitals"],
]

with open("ipl_matches.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(matches)
```

```python
import csv

with open("ipl_matches.csv", "r") as file:
    reader = csv.DictReader(file)
    for row in reader:
        print("Match", row["Match"], "winner:", row["Winner"])
```
